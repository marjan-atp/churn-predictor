```python


columns_to_trim = ['CreditScore', 'Age', 'Balance', 'EstimatedSalary']

df_cleaned = df.copy()

for col in columns_to_trim:
    Q1 = df_cleaned[col].quantile(0.25)
    Q3 = df_cleaned[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    df_cleaned = df_cleaned[(df_cleaned[col] >= lower_bound) &
                            (df_cleaned[col] <= upper_bound)]

# Reset index after dropping rows
df_cleaned = df_cleaned.reset_index(drop=True)

print(f"Original shape : {df.shape}")
print(f"Cleaned shape  : {df_cleaned.shape}")
print(f"Rows removed   : {df.shape[0] - df_cleaned.shape[0]}")

df_cleaned  # ← was showing df before (wrong!)
```

```python
import matplotlib.pyplot as plt
from sklearn.metrics import accuracy_score, ConfusionMatrixDisplay
from sklearn.svm import SVC
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split

# 1. FIRST: Take a 10% sample of your FULL dataset to shrink it down
x_small, _, y_small, _ = train_test_split(
    x, y, train_size=0.10, stratify=y, random_state=42
)

# 2. SECOND: Split that newly shrunk dataset into Train and Test sets
x_train, x_test, y_train, y_test = train_test_split(
    x_small, y_small, test_size=0.20, stratify=y_small, random_state=42
)

# 3. Scale the data (Crucial for SVC speed)
scaler = StandardScaler()
x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)

# 4. Train the model using the SCALED data
svc_m3 = SVC(kernel='linear', C=1.0, random_state=42)
svc_m3.fit(x_train_scaled, y_train)

# 5. Make predictions using the SCALED data
svc_train_pred = svc_m3.predict(x_train_scaled)
svc_test_pred = svc_m3.predict(x_test_scaled)

# 6. Calculate and print accuracies
svc_train_acc = accuracy_score(y_train, svc_train_pred)
svc_test_acc = accuracy_score(y_test, svc_test_pred)

print(f"Train Accuracy: {svc_train_acc:.4f}")
print(f"SVC Test Accuracy: {svc_test_acc:.4f}")
```

```python
# 1. Explicitly enable the experimental feature
from sklearn.experimental import enable_halving_search_cv
from sklearn.model_selection import HalvingGridSearchCV

# Import models
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
import xgboost as xgb

# 2. Initialize the base models
# n_jobs=-1 ensures they use all available processors
models = {
    'KNN': KNeighborsClassifier(n_jobs=-1),
    'RF': RandomForestClassifier(random_state=42, n_jobs=-1),
    # Added tree_method='hist' back to XGB to ensure it doesn't freeze on your massive dataset
    'XGB': xgb.XGBClassifier(tree_method='hist', random_state=42, n_jobs=-1, eval_metric='logloss'),
    # Standard GB integrated here
    'GB': GradientBoostingClassifier(random_state=42)
}

# 3. Define the hyperparameter grids for each model
param_grids = {
    'KNN': {
        'n_neighbors': [55, 63, 79, 100],
        'weights': ['uniform', 'distance'],
        'metric': ['euclidean', 'manhattan']
    },
    'RF': {
        'n_estimators': [30, 50, 80],
        'max_depth': [10, 20, 50],
        # Fixed ValueError trap: min_samples_split must be >= 2
        'min_samples_split': [2, 10, 20],
        'min_samples_leaf': [1, 5, 10],
        'max_features': ['log2', 'sqrt']
    },
    'XGB': {
        'n_estimators': [50, 100, 200, 300],
        'learning_rate': [0.001, 0.01, 0.05, 0.1],
        'max_depth': [3, 5, 7],
        'subsample': [0.8, 1.0],
    },
    # Integrated your params7 for Gradient Boosting
    'GB': {
        'n_estimators': [100, 300],
        'learning_rate': [0.01, 0.1, 0.2],
        'max_depth': [5, 7,9],
        'subsample': [0.8, 1.0]
    }
}

# Dictionary to store the best models after tuning
best_tuned_models = {}

# 4. Iterate through each model and run HalvingGridSearchCV
for model_name in models.keys():
    print(f"--- Tuning {model_name} ---")

    # Initialize the Halving search
    search = HalvingGridSearchCV(
        estimator=models[model_name],
        param_grid=param_grids[model_name],
        factor=3,               # Keeps the top 1/3 of candidates for the next round
        cv=3,                   # 3-fold cross-validation for speed
        scoring='accuracy',
        random_state=42,
        n_jobs=-1,              # Multithread the search process
        verbose=1               # Prints out the progress
    )

    # Fit the search to your training data
    search.fit(x_train, y_train)

    # Save the best version of the model
    best_tuned_models[model_name] = search.best_estimator_

    # Print the results for this model
    print(f"Best Parameters for {model_name}: {search.best_params_}")
    print(f"Best Cross-Validation Score: {search.best_score_:.4f}\n")

print("All tuning complete!")
```

```python
%%writefile app.py
import streamlit as st
import pandas as pd
import joblib

# --- 1. Load the Saved Model ---
try:
    model = joblib.load('best_churn_model.pkl')
except FileNotFoundError:
    st.error("Error: 'best_churn_model.pkl' not found. Please ensure you saved the model in Colab first!")
    st.stop()

# --- 2. Build the Web App UI ---
st.set_page_config(page_title="Churn Predictor", page_icon="🏦", layout="centered")

st.title("🏦 Bank Customer Churn Predictor")
st.write("Enter the customer's details below to predict if they will leave the bank.")
st.markdown("---")

# Create two columns for a cleaner layout
col1, col2 = st.columns(2)

with col1:
    st.subheader("Personal Details")
    Geography = st.selectbox("Geography (Country)", ["France", "Germany", "Spain"])
    Gender = st.selectbox("Gender", ["Male", "Female"])
    Age = st.slider("Age", min_value=18, max_value=100, value=35)
    Tenure = st.slider("Tenure (Years with Bank)", min_value=0, max_value=10, value=5)
    EstimatedSalary = st.number_input("Estimated Salary ($)", min_value=0.0, value=60000.0, step=1000.0)

with col2:
    st.subheader("Account Details")
    CreditScore = st.number_input("Credit Score", min_value=300, max_value=850, value=650)
    Balance = st.number_input("Account Balance ($)", min_value=0.0, value=50000.0, step=1000.0)
    NumOfProducts = st.selectbox("Number of Products", [1, 2, 3, 4])
    HasCrCard = st.selectbox("Has Credit Card?", ["Yes", "No"])
    IsActiveMember = st.selectbox("Is Active Member?", ["Yes", "No"])

st.markdown("---")

# --- 3. Backend Data Conversion ---
# Convert Yes/No to 1/0
has_crcard_encoded = 1 if HasCrCard == "Yes" else 0
is_active_encoded = 1 if IsActiveMember == "Yes" else 0

# Convert Gender to 1/0
# (Assuming Male=1, Female=0 from standard Label Encoding)
gender_encoded = 1 if Gender == "Male" else 0

# Convert Geography to numbers
# (Assuming France=0, Germany=1, Spain=2 from standard Label Encoding)
geo_dict = {"France": 0, "Germany": 1, "Spain": 2}
geography_encoded = geo_dict[Geography]

# --- 4. Make the Prediction ---
if st.button("Predict Churn Risk", type="primary", use_container_width=True):

    # Build the DataFrame in the EXACT order your model was trained on
    input_data = pd.DataFrame({
        'CreditScore': [CreditScore],
        'Geography': [geography_encoded],
        'Gender': [gender_encoded],
        'Age': [Age],
        'Tenure': [Tenure],
        'Balance': [Balance],
        'NumOfProducts': [NumOfProducts],
        'HasCrCard': [has_crcard_encoded],
        'IsActiveMember': [is_active_encoded],
        'EstimatedSalary': [EstimatedSalary]
    })

    # Generate prediction and probability
    prediction = model.predict(input_data)
    probability = model.predict_proba(input_data)[0][1] * 100

    # Display results beautifully
    st.markdown("### Prediction Results:")
    if prediction[0] == 1:
        st.error(f"⚠️ **High Risk of Churn!**")
        st.write(f"The model predicts a **{probability:.1f}%** probability that this customer will close their accounts.")
    else:
        st.success(f"✅ **Safe: Customer is likely to stay.**")
        st.write(f"The model predicts only a **{probability:.1f}%** probability of churn.")
```
